### Exercise 5.1 — BFS on SAD1, SAD2, SAD3, SAfriendofBFS

```python
levels_51 = ["SAD1", "SAD2", "SAD3", "SAfriendofBFS"]
df_51 = df_ex5[
    (df_ex5["Level"].isin(levels_51)) & (df_ex5["Strategy"] == "BFS")
][["Level", "Strategy", "Heuristic", "States Generated", "Time/s", "Solution length"]].reset_index(drop=True)
df_51
```

### Exercise 5.2 — BFS and DFS on SAFirefly and SACrunch

```python
levels_52 = ["SAFirefly", "SACrunch"]
df_52 = df_ex5[
    (df_ex5["Level"].isin(levels_52)) & (df_ex5["Strategy"].isin(["BFS", "DFS"]))
][["Level", "Strategy", "Heuristic", "States Generated", "Time/s", "Solution length"]].reset_index(drop=True)
df_52
```

### Exercise 5.3 — BFS, DFS, and Greedy (GoalCount) on SAFirefly and SACrunch

```python
levels_53 = ["SAFirefly", "SACrunch"]
mask_53 = (
    (df_ex5["Level"].isin(levels_53)) &
    (
        df_ex5["Strategy"].isin(["BFS", "DFS"]) |
        ((df_ex5["Strategy"] == "GREEDY") & (df_ex5["Heuristic"] == "GOALCOUNT"))
    )
)
df_53 = df_ex5[mask_53][
    ["Level", "Strategy", "Heuristic", "States Generated", "Time/s", "Solution length"]
].reset_index(drop=True)
df_53
```

### Exercise 5.4 — Greedy Best-First with Advanced Heuristics on Sokoban levels

```python
soko_levels = [
    "SAsoko1_04", "SAsoko1_08", "SAsoko1_16", "SAsoko1_32", "SAsoko1_64", "SAsoko1_128",
    "SAsoko2_04", "SAsoko2_08", "SAsoko2_16", "SAsoko2_32", "SAsoko2_64", "SAsoko2_128",
    "SAsoko3_04", "SAsoko3_05", "SAsoko3_06", "SAsoko3_07", "SAsoko3_08",
    "SAsoko3_16", "SAsoko3_32", "SAsoko3_64", "SAsoko3_128",
]

# Show ADVANCED and ADVANCEDGOAL side-by-side for easy comparison
df_54 = df_ex5[
    (df_ex5["Level"].isin(soko_levels)) &
    (df_ex5["Strategy"] == "GREEDY") &
    (df_ex5["Heuristic"].isin(["ADVANCED", "ADVANCEDGOAL"]))
][["Level", "Heuristic", "States Generated", "Time/s", "Solution length"]].reset_index(drop=True)

df_54_pivot = df_54.pivot_table(
    index="Level",
    columns="Heuristic",
    values=["States Generated", "Time/s", "Solution length"],
    aggfunc="first"
)
# Flatten multi-level columns
df_54_pivot.columns = [f"{col[0]} ({col[1]})" for col in df_54_pivot.columns]
df_54_pivot = df_54_pivot.reindex(soko_levels).reset_index()
df_54_pivot
```
